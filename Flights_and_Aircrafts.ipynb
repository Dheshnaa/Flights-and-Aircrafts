```python
import sqlalchemy
```

```python
engine = sqlalchemy.create_engine('sqlite:///new.db')
```

```python
con = engine.raw_connection()
```

```python
cursor = con.cursor()
```

```python
import pandas as pd
```

```python
data1 = pd.read_csv('flights-2.csv')
```

```python
data1.to_sql('flights', con, if_exists='append', index=False)
```

```python
for row in cursor.execute("SELECT * FROM flights LIMIT 10;"): print(row)
```

```python
data2 = pd.read_csv('aircrafts.csv')
```

```python
data2.to_sql('aircrafts', con, if_exists='append', index=False)
```

```python
for row in cursor.execute("SELECT * FROM aircrafts LIMIT 10;"): print(row)
```

```python
q_2a =  """SELECT * FROM flights
WHERE Origin = 'LAS'
ORDER BY Depart_Time
ASC LIMIT 5;"""

for row in cursor.execute(q_2a): print(row)
```

```python
q_2b =  """SELECT * FROM flights
WHERE Origin = 'LAS'
ORDER BY Depart_Time
DESC LIMIT 5;"""

for row in cursor.execute(q_2b): print(row)
```

```python
q_3 = """SELECT Origin FROM flights LEFT JOIN aircrafts ON flights.Aircraft = aircrafts.Aircraft
WHERE Seats = 150
GROUP BY Origin
ORDER BY Origin DESC;"""

for row in cursor.execute(q_3): print(row)
```

```python
q_4 = """SELECT aircrafts.Aircraft, COUNT(flights.Flight), SUM(aircrafts.Seats) FROM flights LEFT JOIN aircrafts ON flights.Aircraft = aircrafts.Aircraft
WHERE Origin = 'JFK' AND aircrafts.Aircraft LIKE '%7%'
GROUP BY flights.Aircraft
ORDER BY flights.Aircraft;"""

for row in cursor.execute(q_4): print(row)
```
